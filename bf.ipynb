Étape 2 : Tests Statistiques

```python
print(data["Total_Spending"].describe())
```

1) t-test pour comparer les montants moyens des ventes.

H0 (hypothèse nulle) : il n’y a pas de différence entre les deux groupes.

H1 : il existe une différence.

```python
from scipy.stats import ttest_ind

group1 = data[data["Response"] == 1]["Total_Spending"]
group0 = data[data["Response"] == 0]["Total_Spending"]

t_stat, p_value = ttest_ind(group1, group0)
alpha = 0.05

print("p-value:", p_value)



if p_value < alpha:
    print("Nous rejetons H0 : il existe une différence significative entre les deux groupes (la campagne a un effet).")
else:
    print("Nous ne rejetons pas H0 : il n’y a pas de différence significative (la campagne n’a pas d’effet)")
```

2) teste loi chi2_contingency

```python
from scipy.stats import chi2_contingency
table = pd.crosstab(data["Country"], data["AcceptedCmp1"])
chi2, p, _, _ = chi2_contingency(table)
alpha = 0.05

if p < alpha:
    print("Il y a une relation significative entre le pays et l’acceptation de la campagne")
else:
    print("ne pas Il y a une relation significative entre le pays et l’acceptation de la campagne")
```
